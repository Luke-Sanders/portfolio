## SASS Buttons Homework

In this assignment, I practiced using OCS SASS button classes to create consistent buttons, links, and action layouts without writing custom CSS.



## Popcorn Hack 1: Base Button Class

I added the `ocs__btn` class to the resource directory link so it uses the standard OCS button styling.


```python

%%html
<!-- UI_RUNNER: Buttons Popcorn 1 -->

<a class="ocs__btn" href="https://www.mygoodbrain.org/resources">
  Mental Health Resource Directory
</a>
```


## Popcorn Hack 2: Modifiers and Button Groups

I added the `pill` and `fill` classes to the main action button and used `ocs__links` to organize both actions together.


```python

%%html
<!-- UI_RUNNER: Buttons Popcorn 2 -->

<div class="ocs__links">
  <button type="button" class="ocs__btn pill accent fill">
    Complete Daily Check-In
  </button>

  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources">
    Save Draft Journal
  </a>
</div>
```


## Popcorn Hack 3: Status Tones and Layout

I changed the container to `ocs__links--wide`, updated the check-in button to use `alert-green`, and added `fill` to the crisis support link.


```python

%%html
<!-- UI_RUNNER: Buttons Popcorn 3 -->

<div class="ocs__links--wide">
  <button type="button" class="ocs__btn alert-green outline">
    Complete Check-In
  </button>

  <a class="ocs__btn alert-red fill" href="https://www.mygoodbrain.org/resources">
    Get Crisis Support
  </a>
</div>
```


## Homework Hack 1: Adaptive Resource Toolbar

I created a resource toolbar using OCS SASS classes instead of custom styling.

The three categories use different colors and styles based on their purpose:

- Urgent Support uses a red filled button.
- Self-Care uses a green filled button.
- Community uses an accent outlined button.

All three actions are grouped inside `ocs__links`.


```python

%%html
<!-- UI_RUNNER: Buttons Homework 1 -->

<div class="ocs__links">

  <a
    href="https://www.mygoodbrain.org/resources"
    class="ocs__btn alert-red fill pill">
    Urgent Support
  </a>

  <a
    href="https://www.mygoodbrain.org/resources"
    class="ocs__btn alert-green fill pill">
    Self-Care
  </a>

  <a
    href="https://www.mygoodbrain.org/resources"
    class="ocs__btn accent outline pill">
    Community
  </a>

</div>
```


## Homework Hack 2: Primary vs. Secondary Actions

I created a button group that uses different OCS classes to show which actions are most important.

The main action, Log Mood Entry, uses a filled accent button. View Weekly Trends uses an outline style, while Crisis Hotline uses a red outlined style.

This creates a clear visual hierarchy without needing custom CSS.


```python

%%html
<!-- UI_RUNNER: Buttons Homework 2 -->

<div class="ocs__links">

  <button type="button" class="ocs__btn accent fill">
    Log Mood Entry
  </button>

  <a
    href="https://www.mygoodbrain.org/resources"
    class="ocs__btn outline">
    View Weekly Trends
  </a>

  <a
    href="https://www.mygoodbrain.org/resources"
    class="ocs__btn alert-red outline">
    Crisis Hotline
  </a>

</div>
```


## Reflection

This assignment helped me understand how SASS classes make button styling easier and more consistent. Instead of writing individual CSS rules, I can use classes like `fill`, `outline`, `pill`, and `alert-red`.

I also learned how to organize buttons using `ocs__links` and how different styles can help users recognize which actions are most important.
