## SASS Grids Homework

In this assignment, I practiced using OCS grid classes to organize information into rows and columns without writing custom CSS.



## Popcorn Hack 1: Fix the Beaker Row

I replaced the original layout with OCS grid classes. Each beaker now uses an `ocs__grid-cell` inside a standard grid.


```python

%%html
<!-- UI_RUNNER: popcorn-hack-1 -->
<div class="ocs__grid ocs__grid--standard">
  <div class="ocs__grid-cell">Beaker A: Water</div>
  <div class="ocs__grid-cell">Beaker B: Salt Water</div>
  <div class="ocs__grid-cell">Beaker C: Vinegar</div>
</div>
```


## Popcorn Hack 2: Size the Beakers Using fr

I used four fraction values to create four columns. The control beaker uses `2fr`, making it twice as wide as each of the other beakers.


```python

%%html
<!-- UI_RUNNER: popcorn-hack-2 -->
<div class="ocs__grid ocs__grid--standard" style="grid-template-columns: 1fr 1fr 1fr 2fr;">
  <div class="ocs__grid-cell">Beaker 1</div>
  <div class="ocs__grid-cell">Beaker 2</div>
  <div class="ocs__grid-cell">Beaker 3</div>
  <div class="ocs__grid-cell">Control Beaker</div>
</div>
```


## Homework Hack: Standard Grid and Calculator

For this challenge, I created two grids using OCS classes.

The first grid organizes temperature readings and highlights the outlier beaker. The second uses a calculator grid with an accent cell for the average button.

Both grids use OCS classes instead of custom layout styles.


```python

%%html
<!-- UI_RUNNER: homework-hack -->

<div class="ocs__card">
  <div class="ocs__grid ocs__grid--standard">
    <div class="ocs__grid-cell ocs__grid-cell--header">Temperature Trial (°C)</div>
    <div class="ocs__grid-cell">Beaker 1: 22.4</div>
    <div class="ocs__grid-cell">Beaker 2: 22.9</div>
    <div class="ocs__grid-cell ocs__grid-cell--accent">Beaker 3: 31.6</div>
  </div>
</div>

<div class="ocs__card">
  <div class="ocs__grid ocs__grid--calculator">
    <div class="ocs__grid-cell">7</div>
    <div class="ocs__grid-cell">8</div>
    <div class="ocs__grid-cell">9</div>
    <div class="ocs__grid-cell ocs__grid-cell--accent">Avg</div>
  </div>
</div>
```
