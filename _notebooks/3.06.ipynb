```python
amount_requested = 0

# 1 condition -> 2 outcomes -> 2 answers
if amount_requested > 100:
    print("This message is asking for a large amount of money.")
else:
    print("This message is not asking for a large amount of money.")
```

```python
risk_score = 9

# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
```

## Popcorn Hack 2 answers

**What the broken version told a score of 9:** "Be a little careful, but this is probably fine." That is the wrong advice for the worst possible score.

**Unreachable branches:** `elif risk_score >= 6` and `elif risk_score >= 9`. Any score of 6 or 9 is already caught by `if risk_score >= 3`, so neither branch can run for any score.

**Fix:** order the thresholds from highest to lowest (9, then 6, then 3).

```python
message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False

risk_score = 0

# Each signal is its own if. No else is needed on any of them because
# "signal not present" means add nothing, so risk_score is already correct.

# Gift card signal: if absent, nothing to add
if "gift card" in message:
    risk_score = risk_score + 4

# Large amount signal: if absent, nothing to add
if amount_requested > 100:
    risk_score = risk_score + 3

# Unknown sender signal: if absent, nothing to add
if not known_contact:
    risk_score = risk_score + 2

print("Risk score:", risk_score)

# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
```

```python
message = "Hi Mom, are we still on for Sunday?"
amount_requested = 0
known_contact = True

risk_score = 0

# No else needed: a missing signal adds nothing
if "gift card" in message:
    risk_score = risk_score + 4

if amount_requested > 100:
    risk_score = risk_score + 3

if not known_contact:
    risk_score = risk_score + 2

print("Risk score:", risk_score)

# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
```

## Reachability proof

| Branch | message | amount_requested | known_contact | risk_score | Output |
|---|---|---|---|---|---|
| `if risk_score >= 9` | "send gift cards now" | 500 | False | 4 + 3 + 2 = 9 | Do not reply. This is almost certainly a scam. |
| `elif risk_score >= 6` | "send gift cards now" | 500 | True | 4 + 3 = 7 | This looks risky. Check with someone you trust before replying. |
| `elif risk_score >= 3` | "send gift cards now" | 0 | True | 4 | Be a little careful, but this is probably fine. |
| `else` | "Hi Mom, are we still on for Sunday?" | 0 | True | 0 | Nothing suspicious in this message. |

Every branch has inputs that reach it, so there is no dead code.

## Extreme tests

- **Completely clean message:** score 0, prints "Nothing suspicious in this message."
- **All signals tripped:** score 9, prints "Do not reply. This is almost certainly a scam."

## Stretch

An `if` with no `else` is fine in the scoring step because "signal absent" means "add nothing," so the missing branch is an intended do-nothing. In Example A, the missing branch left the user with no output at all.
