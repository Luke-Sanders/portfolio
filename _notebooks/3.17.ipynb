## Popcorn Hack

```python
import math

n = 8
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1
print(n)
print(linear_checks)
print(binary_checks)
```

```python
import math

n = 16
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1
print(n)
print(linear_checks)
print(binary_checks)
```

Prediction: For 8 books, linear search needs 8 checks in the worst case and binary search needs 4.

Actual output (n = 8): 8, 8, 4.

Actual output (n = 16): 16, 16, 5.

Explanation: When I doubled n from 8 to 16, the linear count doubled (8 to 16), but the binary count only went up by one (4 to 5). Binary search cuts the catalog in half each time, so doubling the catalog only adds one more check.

Recommendation: For a big catalog that is already sorted, use binary search because it needs far fewer checks.

## Homework

Each catalog has n books. Scanning every book takes n visits. Comparing every book with every book takes n * n ordered pairs.

```python
n = 12
visits = n
pairs = n * n
print(n)
print(visits)
print(pairs)
```

Prediction: 12, 12, 144.

Actual output: 12, 12, 144.

```python
n = 24
visits = n
pairs = n * n
print(n)
print(visits)
print(pairs)
```

Prediction: 24, 24, 576.

Actual output: 24, 24, 576.

Explanation of growth: When I doubled the catalog from 12 to 24 books, the visits doubled (12 to 24) but the pairs quadrupled (144 to 576). Visits grow in a straight line with n. Pairs grow much faster because every book is compared with every book, which is n times n.

```python
import math

n = 1000
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1
print(n)
print(linear_checks)
print(binary_checks)
```

Actual output: 1000, 1000, 10.

Recommendation: For repeated searches of a catalog that is already sorted, the librarian should use binary search. It needs at most 10 checks instead of 1,000.

What changes if it is not sorted: Binary search only works on sorted data. If the catalog is unsorted, it has to be sorted first, and sorting takes extra work. For a single search, linear search might be cheaper.

What changes if a sorted copy must be stored: Keeping a second sorted copy uses extra memory, so there is a tradeoff between time and memory.

My goal: Compare ways to search by counting the checks each one needs, instead of judging by how short the code is.

Methods considered: I compared linear search (check one at a time) and binary search (repeatedly halve a sorted catalog).

Prototype: A small calculator that shows the book count, the visits, and the ordered pairs.
