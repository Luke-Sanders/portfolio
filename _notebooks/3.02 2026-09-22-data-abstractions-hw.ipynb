## Popcorn Hack

```python
level_stars = [6, 9, 12]
level_stars[1] = level_stars[1] + 3
level_stars.append(5)
print(level_stars)
print(len(level_stars))
```

Prediction: The list will be [6, 12, 12, 5] and the length will be 4. The AP index for the same second level is 2, because AP lists start at 1.

Actual output: [6, 12, 12, 5], then 4.

Explanation: Replacing an element changes a value but keeps the length the same. Appending adds a new element, so the length goes up by one.

## Homework

Each element of level_stars is one level's collected stars, in level order.

```python
level_stars = [12, 18, 10]

level_stars[1] = level_stars[1] + 5
level_stars.append(7)

total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print(level_stars)
print(len(level_stars))
print(total_stars)
```

Actual output: [12, 23, 10, 7], then 4, then 52.

Explanation: I added 5 to the second level (18 became 23) and appended 7 as a new fourth level. The loop visits every level, so it added up all four without me changing the sum line.

## Tests

### Test 1: Main list

```python
level_stars = [12, 18, 10]
level_stars[1] = level_stars[1] + 5
level_stars.append(7)

total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print(level_stars)
print(len(level_stars))
print(total_stars)
```

Prediction: [12, 23, 10, 7], 4 levels, total 52.

Actual output: [12, 23, 10, 7], 4, 52.

Explanation: The prediction matched because the loop adds every element, including the appended one.

### Test 2: Empty list

```python
level_stars = []

if len(level_stars) > 1:
    level_stars[1] = level_stars[1] + 5

total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print(len(level_stars))
print(total_stars)
```

Prediction: 0 levels and a total of 0.

Actual output: 0, 0.

Explanation: An empty list has no second element, so I skip the update with an if check. Otherwise Python would raise an IndexError. The loop runs zero times, so the total stays at its starting value of 0.

## Design Thinking

Why a list avoids work: With 100 separate variables, I would have to create 100 names and edit the total calculation every time I added a level. With a list, I only append a value and the same loop totals everything.

Player/maker need: A game maker needs to add or reorder levels without losing track of each level's stars.

My goal: Keep every level's stars in one collection that grows.

Representation considered: I considered separate variables like in 3.1, but chose a list because it scales to any number of levels.

Prototype: A list of level stars with an update, an append, and a loop that totals it.

Test and revision: I tested the main list and an empty list. The empty list showed me I needed an if check before updating the second level.

Clear labels: Levels should have names like Forest, Cavern and Summit, so players do not need to understand indexes.
