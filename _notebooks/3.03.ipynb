## Homework: Boss Level Detector

The % operator gives the remainder. If player_level % 5 is 0, the level divides evenly by 5, so it is a boss level.

```python
player_level = 15

is_boss_level = player_level % 5

if is_boss_level == 0:
    print("WARNING: Level " + str(player_level) + " is a Boss Level! Prepare")
else:
    print("Level " + str(player_level) + " cleared! Regular enemies ahead.")
```

## Homework: Game Score & Bonus Tracker

Python does multiplication before addition (PEMDAS), so 50 * 2 + 15 is 100 + 15 = 115. The % 25 check shows how far the score is from a multiple of 25, which earns an extra life.

```python
base_score = 50
multiplier = 2
bonus_points = 15

total_score = base_score * multiplier + bonus_points
print("Total Score: " + str(total_score))

remainder = total_score % 25
if remainder == 0:
    print("Bonus! You earned an extra life!")
else:
    print("Points needed for next extra life: " + str(25 - remainder))
```

### My own version: adding a power-up

I added a powerup that gives 10 extra points. Now the score is 50 * 2 + 15 + 10 = 125, which divides evenly by 25, so I earn an extra life.

```python
base_score = 50
multiplier = 2
bonus_points = 15
power_up = 10

total_score = base_score * multiplier + bonus_points + power_up
print("Total Score: " + str(total_score))

remainder = total_score % 25
if remainder == 0:
    print("Bonus! You earned an extra life!")
else:
    print("Points needed for next extra life: " + str(25 - remainder))
```
